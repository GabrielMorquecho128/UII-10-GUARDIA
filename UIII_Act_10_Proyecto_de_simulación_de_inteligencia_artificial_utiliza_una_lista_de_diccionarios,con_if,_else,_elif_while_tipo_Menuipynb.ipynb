```python
# ===================================
# 🛡️ GUARDIA IA LISTA DE DICCIONARIOS
# ===================================

# Lista vacia
guardias = []

# Diccionario vacio
losguardias = {}

print("===================================")
print("   GUARDIA IA LISTA DICCIONARIOS")
print("===================================")

while True:

    print("\n=========== MENU ===========")
    print("1. Agregar guardia")
    print("2. Mostrar guardias")
    print("3. Analizar guardia")
    print("4. Salir")

    opcion = int(input("Seleccione una opcion: "))

    # -----------------------------------
    # OPCION 1
    # -----------------------------------

    if opcion == 1:

        nombre = input("Ingrese nombre del guardia: ")

        zona = input("Ingrese zona de vigilancia: ")

        horas = int(input("Ingrese horas trabajadas: "))

        incidentes = int(input("Ingrese cantidad de incidentes: "))

        losguardias = {

            "Nombre": nombre,
            "Zona": zona,
            "Horas": horas,
            "Incidentes": incidentes

        }

        guardias.append(losguardias)

        print("\nGuardia agregado correctamente")

    # -----------------------------------
    # OPCION 2
    # -----------------------------------

    elif opcion == 2:

        print("\n==============================================")
        print("         LISTA DE GUARDIAS")
        print("==============================================")

        if len(guardias) == 0:

            print("No hay guardias registrados")

        else:

            for datos in guardias:

                print("Nombre      :", datos["Nombre"])
                print("Zona        :", datos["Zona"])
                print("Horas       :", datos["Horas"])
                print("Incidentes  :", datos["Incidentes"])
                print("----------------------------------------------")

    # -----------------------------------
    # OPCION 3
    # -----------------------------------

    elif opcion == 3:

        incidentes = int(input("Ingrese cantidad de incidentes: "))

        acceso = input("¿Tiene acceso autorizado? (si/no): ")

        print("\nANALIZANDO GUARDIA...\n")

        if incidentes >= 5 and acceso == "no":

            print("Zona de riesgo detectada.")
            print("Enviar apoyo inmediato.")

        elif incidentes >= 1 and incidentes <= 4:

            print("Incidentes menores detectados.")
            print("Mantener vigilancia.")

        else:

            print("Guardia estable.")
            print("Sin riesgos detectados.")

    # -----------------------------------
    # OPCION 4
    # -----------------------------------

    elif opcion == 4:

        print("\n===================================")
        print("      PROGRAMA FINALIZADO")
        print("===================================")

        print("Alumno: Marco Morquecho")
        print("Especialidad: Inteligencia Artificial")
        print("Materia: Elabora Proyectos de Programacion Logica")
        print("Docente: Eliseo Nava")
        print("CBTis 128")

        print("\nGracias por utilizar GUARDIA IA")

        break

    # -----------------------------------
    # OPCION INVALIDA
    # -----------------------------------

    else:

        print("Opcion no valida")
```
